### **Załadowanie bibliotek**

```python
import pandas as pd
```

### **Wczytanie danych**

```python
plik_1 = pd.read_excel('1-500000_note_loan_book.xlsx')
```

```python
plik_2 = pd.read_excel('500001-1000000_note_loan_book.xlsx')
```

```python
plik_3 = pd.read_excel('1000001-1500000_note_loan_book.xlsx')
```

**Usunięcie rekordów, dla których "Loan Status" = 'Current', 'Late 1-15', 'Late 16-30', 'Late 31-60', 'Grace Period', 'Late 60+'**

```python
plik_1 = plik_1[~plik_1["Loan Status"].isin(["Current", "Late 1-15", "Late 16-30", "Late 31-60", "Grace Period", "Late 60+"])]
```

```python
len(plik_1)
```

```python
plik_2 = plik_2[~plik_2["Loan Status"].isin(["Current", "Late 1-15", "Late 16-30", "Late 31-60", "Grace Period", "Late 60+"])]
```

```python
len(plik_2)
```

```python
plik_3 = plik_3[~plik_3["Loan Status"].isin(["Current", "Late 1-15", "Late 16-30", "Late 31-60", "Grace Period", "Late 60+"])]
```

```python
len(plik_3)
```

### **Połączenie danych w jeden DataFrame**

```python
dane = pd.concat([plik_1, plik_2, plik_3], ignore_index=True)
```

```python
print(dane.shape)
```

```python
dane.to_csv("dane.csv", index=False)
```
